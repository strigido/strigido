```python
from numpy import linspace
def generate(func, xs: list[float]):
    for x in xs:
        y = func(x)
        print(f"{{ x: {x}, y: {y} }},", end="")
```

### Odvisnost lomilne sile od števila špagetov v snopu

```python
generate (
    lambda x: 0.773*x+2.775,
    list(range(10, 51, 1))
)
```

### Odvisnost lomilne sile od debeline špagetov

```python
generate (
    lambda x: 2.2814*x-2.0886,
    list(linspace(1.1, 1.9, 50))
)
```
