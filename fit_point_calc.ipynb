```python
from numpy import linspace
def generate(func, xs: list[float]):
    for x in xs:
        y = func(x)
        print(f"{{ x: {x}, y: {y} }},", end="")
```

### Odvisnost lomilne sile od števila špagetov v snopu

```python
generate (
    lambda x: 0.773*x+2.775,
    list(range(10, 51, 1))
)
```

### Odvisnost lomilne sile od debeline špagetov

```python
generate (
    lambda x: 2.2814*x-2.0886,
    list(linspace(1.1, 1.9, 50))
)
```

### Odvisnost lomilne sile od razdalje med podporama

```python
generate (
    lambda x: 6.5098*x**-0.812,
    list(linspace(4, 20, 50))
)
```

### Odvisnost lomne napetosti od debeline špagetov

```python
generate (
    lambda x: -13.819*x+203.62,
    list(linspace(1.1, 1.9, 50))
)
```

### Odvisnost lomne napetosti od razdalje med podporama

```python
generate (
    lambda x: 2.9159*x+126.29,
    list(linspace(4, 20, 16*2+1))
)
```

### Odvisnost ukrivljenosti od debeline špagetov

```python
generate (
    lambda x: 3.5341*x**-0.77,
    list(linspace(1.1, 1.9, 50))
)
```

### Odvisnost ukrivljenosti od razdalje med podporama

```python
generate (
    lambda x: 0.0213*x**1.6898,
    list(linspace(7.5, 20, 25+1))
)
```
